# First steps

A notebook that exercises what the converter handles: math, printed output, an image, an HTML table and the cell tags.

The grid spacing is $h = (b - a)/n$, and the midpoint rule reads

$$
\int_a^b f(x)\,dx \approx h \sum_{i=1}^{n} f(x_{i-1/2}).
$$

Braces {like these} and a stray < are escaped for MDX.

```python
import numpy as np
from example_pkg import Grid, linspace_midpoints

grid = Grid(0.0, np.pi, 8)
print(grid.spacing)
print(grid.spacing * np.sin(linspace_midpoints(0.0, np.pi, 8)).sum())
```

```python
print(grid)
print(grid.nodes())
```

```python
import matplotlib.pyplot as plt

x = grid.nodes()
fig, ax = plt.subplots(figsize=(5, 2.5))
ax.plot(x, np.sin(x), "o-")
ax.set_xlabel("x")
plt.show()
```

```python
from IPython.display import HTML

HTML(
    "<table><tr><th>cells</th><th>spacing</th></tr>"
    + "".join(f"<tr><td>{n}</td><td>{np.pi / n:.4f}</td></tr>" for n in (4, 8, 16))
    + "</table>"
)
```
